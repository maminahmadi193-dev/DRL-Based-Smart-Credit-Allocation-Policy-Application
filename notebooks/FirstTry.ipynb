## Feature Engineering

```python
def process_data(json_str):

    new_col_names = [  
        'DebtAmount', 'OriginAmount', 'LoanCounts','SarResid', 'Moavagh', 'Mashkuk', 'DefaultFlag', 'DebtRatio', 'DebtPerLoan'
    ]

    try:
        data = json.loads(json_str)
        loan_rows = data.get('Data', {}).get('EstelamAsliRows', [])
        if not loan_rows:
            return pd.Series([0]*len(new_col_names), index=  new_col_names)
        for row in loan_rows:
            for key in ['AmBedehiKol', 'AmOriginal', 'AmSarResid', 'AmMoavagh', 'AmMashkuk']:
                row[key] = float(row.get(key, 0)) if row.get(key) is not None else 0.0

        debt_amount = sum(row["AmBedehiKol"] for row in loan_rows)
        origin_amount = sum(row["AmOriginal"] for row in loan_rows)
        loan_counts = len(loan_rows)

        default_flag = 0
        total_penalty_points= 0
        weighted_penalty_numerator= 0
        defaulted_loans_count = 0
        debt_ratio_list = []


        for row in loan_rows:
            sarresid_val = float(row.get('AmSarResid', 0) or 0)
            moavagh_val = float(row.get('AmMoavagh', 0) or 0)
            mashkuk_val = float(row.get('AmMashkuk', 0) or 0) 

            pi = (1 * sarresid_val) + (3 * moavagh_val) + (60 * mashkuk_val)

            if pi > 0:
                default_flag = 1 
                defaulted_loans_count +=1

            if row['AmOriginal'] > 0:
                debt_ratio_list.append(row['AmBedehiKol']/row['AmOriginal'])

            total_penalty_points+=pi
            weighted_penalty_numerator+= pi*row['AmBedehiKol']


        debt_ratio = np.mean(debt_ratio_list) if debt_ratio_list else 0 
        debt_per_loan = debt_amount/loan_counts if loan_counts > 0 else 0

        return pd.Series([
        debt_amount, origin_amount, loan_counts, sarresid_val, moavagh_val, mashkuk_val, default_flag,
        debt_ratio, debt_per_loan,
    ], index=new_col_names)
    except(json.JSONDecodeError, TypeError, AttributeError):
        return pd.Series([np.nan] * len(new_col_names), index=new_col_names)
```

```python
class CustomCreditEnv(gym.Env):
    """
    محیط یادگیری تقویتی برای تخصیص ریسک اعتباری
    """
    def __init__(self, features, labels, int_rate, raw_incomes, raw_debts, penalty_rate):
        super(CustomCreditEnv, self).__init__()
        
        # داده‌های ورودی
        self.features = np.array(features, dtype=np.float32)
        self.labels = np.array(labels, dtype=np.float32)
        self.n_samples = len(self.features)
        self.raw_incomes = raw_incomes
        self.raw_debts = raw_debts

        self.max_steps_per_episode = 1000  # محدود کردن طول هر اپیزود
        self.steps_in_current_episode = 0
        # پارامترهای مالی پاداش
        self.int_rate = int_rate           # نرخ سود وام
        self.penalty_rate = penalty_rate   # نرخ جریمه سوخت شدن اصل پول
        
        # نگاشت اکشن‌های گسسته به مبالغ واقعی وام (به میلیون تومان)
        self.action_mapping = {
            0: 0.0,
            1: 50.0,
            2: 100.0,
            3: 200.0,
            4: 500.0
        }
        
        # فضای اکشن: ۵ حالت گسسته
        self.action_space = spaces.Discrete(len(self.action_mapping))
        
        # فضای حالت: برداری از ویژگی‌های پیوسته و باینری
        n_features = self.features.shape[1]
        self.observation_space = spaces.Box(
            low=-np.inf, high=np.inf, shape=(n_features,), dtype=np.float32
        )
        
        # وضعیت فعلی در دیتاست
        self.current_step = 0
        self.indices = np.arange(self.n_samples)

    def reset(self, seed=None, options=None):
        super().reset(seed=seed)
        
        # در شروع هر اپیزود، ترتیب مشتریان را بر هم می‌زنیم (Shuffle)
        np.random.shuffle(self.indices)
        self.current_step = 0
        self.steps_in_current_episode = 0
        return self.features[self.indices[self.current_step]], {}

    def step(self, action):
        idx = self.indices[self.current_step]
        client_default = self.labels[idx]
        loan_amount = self.action_mapping[action]
        
        income = self.raw_incomes[idx]
        total_debts = self.raw_debts[idx]
        
        dti = 0.0 # مقدار اولیه برای اطلاعات خروجی (info)
        
        # ----------------------------
        # محاسبه تابع پاداش (Reward)
        # ----------------------------
        if loan_amount == 0.0:
            reward = 0.0
        else:
            # فرض می‌کنیم وام‌ها میانگین ۳۶ ماهه هستند
            existing_installment = total_debts / 36.0
            new_installment = loan_amount / 36.0
            
            # جلوگیری از خطای تقسیم بر صفر برای درآمدهای صفر
            safe_income = max(income, 1.0)
            
            # محاسبه DTI (نسبت مجموع اقساط به درآمد ماهانه)
            dti = (existing_installment + new_installment) / safe_income

            dti = min(dti, 1.5) # قفل کردن DTI روی سقف منطقی برای جلوگیری از انفجار جریمه
            
            # نرمال‌سازی مبلغ وام به واحدهای ۱۰۰ میلیونی برای محاسبات سهمی‌وار (0.5, 1, 2, 5)
            # این کار باعث می‌شود L^2 برای وام ۵۰۰ میلیونی عدد بزرگ ۲۵ شود!
            L = loan_amount / 100.0 
            
            if client_default == 1:
                # ۱. مشتری بدحساب (نکول): جریمه توانی سنگین
                # w_default = 10.0
                # فرمول: - (مبلغ_نرمال^2) * ضریب_نکول * نرخ_جریمه
                # reward = - (L ** 2) * w_default * self.penalty_rate
                reward = - (L) * self.penalty_rate

                
            else:
                # ۲. مشتری خوش‌حساب: موازنه سود خطی و ظرفیت مالی
                w_profit = 20
                w_capacity = 2.75
                
                # سود خطی: متناسب با مبلغ وام
                # profit = L * w_profit * self.int_rate
                profit = L * self.int_rate

                
                # جریمه ظرفیت مالی: توانی نسبت به مبلغ وام، ضربدر نسبتِ DTI
                # هرچه وام بزرگتر یا DTI بیشتر باشد، شیب جریمه وحشتناک بالا می‌رود
                # capacity_penalty = (L ** 2) * dti * w_capacity
                capacity_penalty = (L) * dti 

                
                reward = profit - capacity_penalty
                
                # جریمه بازدارنده ثانویه برای وام‌هایی که از منطق اقتصادی خارجند
                if dti > 0.6:
                    # reward -= 20.0  # جریمه ثابت اضافی برای DTI خطرناک
                    reward -= 5.0

            MAX_REWARD = 2
            reward = np.clip(reward / MAX_REWARD, -1.0, 1.0)

        # ----------------------------
        # انتقال به حالت بعد (Transition)
        # ----------------------------
        self.current_step += 1
        self.steps_in_current_episode += 1

        terminated = self.current_step >= len(self.features)
        truncated = self.steps_in_current_episode >= self.max_steps_per_episode
        
        if terminated or truncated:
            next_obs = self.features[0] 
        else:
            next_obs = self.features[self.indices[self.current_step]]
            
        info = {'DTI': dti, 'Loan': loan_amount, 'Default': client_default} 
        return next_obs, float(reward), terminated, truncated, info
```

```python
# تبدیل دیتافریم‌های نهایی شما به آرایه نامپای برای محیط
features_array = df_final.values
labels_array = reward_data.values

# ساخت محیط با داده‌های دقیق شما
env = CustomCreditEnv(
    features=features_array, 
    labels=labels_array,
    raw_incomes = raw_incomes_array,
    raw_debts= raw_debts_array,
    int_rate=0.23,       # سود ۲۰ درصد (قابل تغییر)
    penalty_rate=1.0     # جریمه برابر با کل مبلغ وام (قابل تغییر)
)

print(f"Environment created successfully!")
print(f"Observation Space Shape: {env.observation_space.shape}")
print(f"Number of Samples: {env.n_samples}")
```

### DQN

```python
from stable_baselines3 import DQN
from stable_baselines3.common.evaluation import evaluate_policy
from stable_baselines3.common.monitor import Monitor

# ۱. رفع هشدار: استفاده از Monitor برای محیط ارزیابی
eval_env = Monitor(env)

model = DQN(
    "MlpPolicy", 
    env, 
    verbose=1,                 # تغییر به 1 تا لاگ آموزش (Loss) را ببینید
    learning_rate=1e-4, 
    buffer_size=50000, 
    learning_starts=1000,      # <--- کلید حل مشکل: آموزش زودتر شروع شود
    batch_size=64,             # آپدیت مکرر با بچ‌های کوچک
    gamma=0.0,                 # <--- چون اکشن روی مشتری بعدی اثر ندارد
    exploration_fraction=0.1, 
    device='auto'
)

# ۲. افزایش زمان آموزش (مثلا 100,000 گام که حداقل یک دور کامل روی داده‌ها بزند)
print("شروع آموزش مدل (ممکن است کمی طول بکشد)...")
model.learn(total_timesteps=100000) 
print("آموزش پایان یافت.")

# ۳. ارزیابی مدل با محیط مانیتور شده
mean_reward, std_reward = evaluate_policy(model, eval_env, n_eval_episodes=50)
print(f"Mean reward: {mean_reward} +/- {std_reward}")

# ۴. بررسی رفتار مدل (ببینیم همیشه صفر میده یا نه)
print("\n--- تست رفتار مدل روی 10 مشتری تصادفی ---")
obs, info = env.reset()
action_counts = {0: 0, 1: 0, 2: 0, 3: 0, 4: 0} # اگر ۵ سطح وام دارید، دیکشنری را کامل کنید

for i in range(10):
    # گرفتن اکشن از مدل آموزش دیده
    action, _states = model.predict(obs, deterministic=True)
    
    # تبدیل اکشن از فرمت نامپای به عدد صحیح پایتون
    action_scalar = int(action.item()) if hasattr(action, 'item') else int(action)
    
    action_counts[action_scalar] += 1
    
    # اجرای اکشن در محیط (ارسال عدد صحیح)
    obs, reward, terminated, truncated, info = env.step(action_scalar)
    print(f"Customer {i+1}: Action Taken = {action_scalar}, Reward = {reward}")
    
    if terminated or truncated:
        obs, info = env.reset()

print(f"\nخلاصه اکشن‌های تست: {action_counts}")
```
