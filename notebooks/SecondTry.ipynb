## Feature Engineering

```python
def process_data(json_str):

    new_col_names = [  
        'DebtAmount', 'OriginAmount', 'LoanCounts','SarResid', 'Moavagh', 'Mashkuk', 'DefaultFlag', 'DebtRatio', 'DebtPerLoan'
    ]

    try:
        data = json.loads(json_str)
        loan_rows = data.get('Data', {}).get('EstelamAsliRows', [])
        if not loan_rows:
            return pd.Series([0]*len(new_col_names), index=  new_col_names)
        for row in loan_rows:
            for key in ['AmBedehiKol', 'AmOriginal', 'AmSarResid', 'AmMoavagh', 'AmMashkuk']:
                row[key] = float(row.get(key, 0)) if row.get(key) is not None else 0.0

        debt_amount = sum(row["AmBedehiKol"] for row in loan_rows)
        origin_amount = sum(row["AmOriginal"] for row in loan_rows)
        loan_counts = len(loan_rows)

        default_flag = 0
        total_penalty_points= 0
        weighted_penalty_numerator= 0
        defaulted_loans_count = 0
        debt_ratio_list = []


        for row in loan_rows:
            sarresid_val = float(row.get('AmSarResid', 0) or 0)
            moavagh_val = float(row.get('AmMoavagh', 0) or 0)
            mashkuk_val = float(row.get('AmMashkuk', 0) or 0) 

            pi = (1 * sarresid_val) + (3 * moavagh_val) + (60 * mashkuk_val)

            if pi > 0:
                default_flag = 1 
                defaulted_loans_count +=1

            if row['AmOriginal'] > 0:
                debt_ratio_list.append(row['AmBedehiKol']/row['AmOriginal'])

            total_penalty_points+=pi
            weighted_penalty_numerator+= pi*row['AmBedehiKol']


        debt_ratio = np.mean(debt_ratio_list) if debt_ratio_list else 0 
        debt_per_loan = debt_amount/loan_counts if loan_counts > 0 else 0

        return pd.Series([
        debt_amount, origin_amount, loan_counts, sarresid_val, moavagh_val, mashkuk_val, default_flag,
        debt_ratio, debt_per_loan,
    ], index=new_col_names)
    except(json.JSONDecodeError, TypeError, AttributeError):
        return pd.Series([np.nan] * len(new_col_names), index=new_col_names)
```

### Build MDP model

```python
import gymnasium as gym
import numpy as np

class LoanEnv(gym.Env):

    def __init__(self, df):
        self.df = df.reset_index(drop=True)
        # دقت کنید OriginAmount اسکیل شده است، پس برای unsecured_amount بهتر است از مقدار منطقی استفاده کنیم
        # اگر می‌خواهید مقدار واقعی باشد، بهتر است OriginAmount خام را هم در دیتافریم پاس دهید
        self.actions = np.array([0.0, 0.25, 0.50, 0.75, 1.0])
        self.action_space = gym.spaces.Discrete(5)
        
        # استخراج مینیمم و ماکزیمم داده‌های اسکیل‌شده برای برگرداندن به بازه [0, 1] در تابع پاداش
        self.risk_min = self.df["Risk_Max"].min()
        self.risk_max = self.df["Risk_Max"].max()
        
        self.score_min = self.df["TotalCustomerScore"].min()
        self.score_max = self.df["TotalCustomerScore"].max()
        
        self.debt_min = self.df["DebtRatio"].min()
        self.debt_max = self.df["DebtRatio"].max()

        self.state_columns = [
            "MountlyIncome", "IncomeScore", "JobScore", "AssetScore",
            "TotalCustomerScore", "TotalFacilities", "DebtAmount",
            "OriginAmount", "LoanCounts", "DebtRatio", "DebtPerLoan",
            "Has_No_History", "Score_Max", "Score_Min", "Risk_Max",
            "Risk_Min", "Employed"
        ]

        self.observation_space = gym.spaces.Box(
            low=-np.inf,
            high=np.inf,
            shape=(len(self.state_columns),),
            dtype=np.float32
        )

    def reset(self, seed=None, options=None):
        super().reset(seed=seed)
        self.idx = np.random.randint(len(self.df))
        row = self.df.iloc[self.idx]
        state = row[self.state_columns].values.astype(np.float32)
        return state, {}
    
    def step(self, action):
        row = self.df.iloc[self.idx]
        ratio = self.actions[action]
        
        # ---------------- محاسبه پاداش ----------------
        if row["DefaultFlag"] == 1:
            reward = -10.0 * ratio
            h_s = None
        else:
            if row["Has_No_History"] == 1:
                h_s = 0.5 
            else:
                # تبدیل مقادیر اسکیل شده (StandardScaler) به بازه 0 تا 1
                norm_risk = (row["Risk_Max"] - self.risk_min) / (self.risk_max - self.risk_min + 1e-8)
                norm_score = (row["TotalCustomerScore"] - self.score_min) / (self.score_max - self.score_min + 1e-8)
                norm_debt = (row["DebtRatio"] - self.debt_min) / (self.debt_max - self.debt_min + 1e-8)
                
                w1, w2, w3 = 0.4, 0.4, 0.2
                
                # امتیاز بالاتر = ریسک کمتر، بنابراین (1 - norm_score)
                h_s = w1 * norm_risk + w2 * (1.0 - norm_score) + w3 * norm_debt
                h_s = np.clip(h_s, 0.0, 1.0) 

            lam = 5
            # reward = ratio * (1.0 - lam * h_s)
            reward = ratio - (lam * h_s * ratio**2)
        # ----------------------------------------------------

        next_state = np.zeros(len(self.state_columns), dtype=np.float32)
        terminated = True
        truncated = False

        info = {
            "ratio": ratio,
            "default": row["DefaultFlag"],
            "H_s": h_s,
            "reward": reward
        }

        return next_state, reward, terminated, truncated, info
```

### DQN

```python
from stable_baselines3 import DQN
from stable_baselines3.common.evaluation import evaluate_policy
from stable_baselines3.common.monitor import Monitor

env = LoanEnv(df_final)

# ۱. رفع هشدار: استفاده از Monitor برای محیط ارزیابی
eval_env = Monitor(env)

model = DQN(
    "MlpPolicy", 
    env, 
    verbose=1,                 # تغییر به 1 تا لاگ آموزش (Loss) را ببینید
    learning_rate=1e-4, 
    buffer_size=50000, 
    learning_starts=1000,      # <--- کلید حل مشکل: آموزش زودتر شروع شود
    batch_size=64,             # آپدیت مکرر با بچ‌های کوچک
    gamma=0.0,                 # <--- چون اکشن روی مشتری بعدی اثر ندارد
    exploration_fraction=0.1, 
    device='auto'
)

# ۲. افزایش زمان آموزش (مثلا 100,000 گام که حداقل یک دور کامل روی داده‌ها بزند)
print("شروع آموزش مدل (ممکن است کمی طول بکشد)...")
model.learn(total_timesteps=100000) 
print("آموزش پایان یافت.")

# ۳. ارزیابی مدل با محیط مانیتور شده
mean_reward, std_reward = evaluate_policy(model, eval_env, n_eval_episodes=50)
print(f"Mean reward: {mean_reward} +/- {std_reward}")

# ۴. بررسی رفتار مدل (ببینیم همیشه صفر میده یا نه)
print("\n--- تست رفتار مدل روی 10 مشتری تصادفی ---")
obs, info = env.reset()
action_counts = {0: 0, 1: 0, 2: 0, 3: 0, 4: 0} # اگر ۵ سطح وام دارید، دیکشنری را کامل کنید

for i in range(10):
    # گرفتن اکشن از مدل آموزش دیده
    action, _states = model.predict(obs, deterministic=True)
    
    # تبدیل اکشن از فرمت نامپای به عدد صحیح پایتون
    action_scalar = int(action.item()) if hasattr(action, 'item') else int(action)
    
    action_counts[action_scalar] += 1
    
    # اجرای اکشن در محیط (ارسال عدد صحیح)
    obs, reward, terminated, truncated, info = env.step(action_scalar)
    print(f"Customer {i+1}: Action Taken = {action_scalar}, Reward = {reward}")
    
    if terminated or truncated:
        obs, info = env.reset()

print(f"\nخلاصه اکشن‌های تست: {action_counts}")
```

```python
import numpy as np
from collections import Counter
import matplotlib.pyplot as plt

# فرض بر این است که مدل شما model و محیط شما env نام دارد
# و env.df شامل دیتاست اسکیل شده شماست.

action_counts = Counter()
all_actions = []

# ریست کردن محیط برای اطمینان
obs, _ = env.reset()

# اجرای مدل روی تمام داده‌های محیط
for i in range(len(env.df)):
    # دریافت اکشن از مدل (بدون اکسپلوریشن)
    action, _states = model.predict(obs, deterministic=True)
    
    action_counts[action.item()] += 1
    all_actions.append(action.item())
    
    # رفتن به استیت بعدی
    obs, reward, terminated, truncated, info = env.step(action)
    
    # چون اپیزودها تک مرحله‌ای هستند، محیط را ریست می‌کنیم تا به داده بعدی برود
    if terminated or truncated:
        obs, _ = env.reset()

# چاپ نتایج
print("توزیع اکشن‌ها در کل دیتاست:")
for act, count in sorted(action_counts.items()):
    ratio = env.actions[act]
    print(f"Action {act} (Ratio: {ratio*100}%): {count} times")

# رسم نمودار
plt.bar(action_counts.keys(), action_counts.values())
plt.xlabel('Action (Loan Ratio)')
plt.ylabel('Count')
plt.title('Action Distribution over Entire Dataset')
plt.xticks(list(action_counts.keys()), [f"{env.actions[a]*100}%" for a in action_counts.keys()])
plt.show()
```
